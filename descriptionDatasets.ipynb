```python
def calculateDescriptiveStatisticsLogs(log):
    no_cases=len(log["case:concept:name"].unique())
    
    no_unique_activities=len(log["concept:name"].unique())
    
    size_cases=log.groupby("case:concept:name").apply(lambda x: len(x["concept:name"]))
    avg_size_cases=np.mean(size_cases)

    variants = pm4py.get_variants(
    log,
    activity_key='concept:name',
    case_id_key='case:concept:name',
    timestamp_key='time:timestamp'
    )
    no_variants=len(list(variants.keys()))

    description_log={}
    description_log["no_cases"]=no_cases
    description_log["no_unique_acts"]=no_unique_activities
    description_log["avg_size_cases"]=avg_size_cases
    description_log["no_variants"]=no_variants

    return description_log
```

```python
rtfm_dataset["Class"].value_counts()/sum(rtfm_dataset["Class"].value_counts())*100
```

```python
calculateDescriptiveStatisticsLogs(sepsis_log)
```
